For a non-negative integer duration, return `(hours, remaining_minutes)`. Unpack one result into two variables.


```python
def split_duration(minutes):
    if type(minutes) == int and not minutes < 0:
        hours = minutes // 60
        remaining_minutes = minutes % 60
        return (hours, remaining_minutes)
    else:
        return "Must enter a non-negative integer!"


# TODO

# Examples: uncomment after completing the exercise.
print(split_duration(95))  # Expected: (1, 35)
print(split_duration(60))  # Expected: (1, 0)
```

### 8.4

Return `(total, average)` for a list of numbers. For an empty list, return `(0, None)`.


```python
def summarise(values):
    total = 0
    if values:
        for i in values:
            total += i
        average = total/len(values)
        return (total, average)
    else:
        return(0, None)

# TODO

# Examples: uncomment after completing the exercise.
print(summarise([2, 0, 4]))  # Expected: (6, 2.0)
print(summarise([]))  # Expected: (0, None)
```

### 8.5

Return a sorted list of unique values in `required` that are absent from `available`. Use sets.


```python
def find_missing(required, available):
    missing = []
    for i in required:
        if i not in available and i not in missing:
            missing.append(i)
    missing.sort()
    return missing

# TODO

# Examples: uncomment after completing the exercise.
print(find_missing([3, 1, 3, 5], [1, 2]))  # Expected: [3, 5]
print(find_missing([], [1]))  # Expected: []
```

### 8.6

Complete the helpers and dispatcher:

- `difference(a, b)`: return `a - b`.
- `percentage(a, b)`: return `a / b * 100`, or None if `b == 0`.
- `calculate(a, b, operation)`: call the matching helper for `"difference"` or `"percentage"`; return None for other operations or non-numeric inputs. Assume booleans are not supplied.


```python
def difference(a, b):
    return a - b


def percentage(a, b):
    if b:
        return (a / b) * 100
    else:
        return None


def calculate(a, b, operation):
    if not (operation in ("difference", "percentage") and type(a) in (int, float) and type(b) in (int, float)):
        return None
    elif operation == "percentage":
        return percentage(a, b)
    else:
        return difference(a, b)

# TODO

# Examples: uncomment after completing the exercise.
print(calculate(15, 20, 'difference'))  # Expected: -5
print(calculate(15, 20, 'percentage'))  # Expected: 75.0
print(calculate(1, 0, 'percentage'))  # Expected: None
print(calculate('1', 2, 'difference'))  # Expected: None
print(calculate(1, 2, 'other'))  # Expected: None
```
